```python
lst = [1, 2]
lst.append(3)
print(lst)

lst.insert(1, 'a')
print(lst)

lst.extend([4, 5])
print(lst)

lst.remove(1)
print(lst)

lst.pop(1)
print(lst)
lst.pop()
print(lst)

lst.clear()
print(lst)
```

```python
list1 = list(range(5))
list2 = list(range(1, 7))
list3 = list(range(0, 5, 2))
print(list1)
print(list2)
print(list3)
print('최솟값, 최댓값 :', min(list1), max(list1))
print('합계 :', sum(list1))
```

```python
lst = [40, 20, 30, 50, 20]
print(lst.index(20))
print(lst.index(20, 2))

print(lst.count(20))
print(lst.count('a'))

lst.sort()
print(lst)
lst.sort(reverse = True)
print(lst)

lst.reverse()
print(lst)
lst.reverse()
print(lst)

list1 = [1, [2,3], 4]
list2 = list1.copy()
print(list1)
print(list2)
list1[0] = 10
list1[1][0] = 20
print(list1)
print(list2)
```
