```python
a = int(input('정수 입력하세요'))
if a%2 == 0:
  print('짝수')
else:
  print('홀수')
```

```python
a = int(input('년도 입력하세요'))
if (a%4 == 0 and a%100!= 0 or a%400 == 0):
  print('윤년')
else:
  print('윤년X')
```

```python
a = int(input('나이를 입력하세요'))
if (a <= 18):
  print('청소년/어린이 교통 카드가 발급됩니다. 교통 요금이 할인됩니다.')
else:
  print('성인 교통 카드가 발급됩니다. 일반 교통 요금이 적용됩니다.')
```

```python
a = int(input('월을 입력하세요'))
if (a >= 3 and a <= 5):
  print('봄')
elif (a >= 6 and a <= 8):
  print('여름')
elif (a >= 9 and a <= 11):
  print('가을')
elif (a > 12 or a <= 0):
  print('입력이 잘못 되었습니다.')
else:
  print('겨울')
```

```python
a = int(input('점수를 입력하세요'))
if (a >= 90):
  print('A')
elif (a >= 80 and a < 90):
  print('B')
elif (a >= 70 and a < 80):
  print('C')
elif (a >= 60 and a < 70):
  print('D')
else:
  print('F')
```
